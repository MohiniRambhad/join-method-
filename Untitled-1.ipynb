```python
import pandas as pd

data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P106", "P107", "P108", "P109", "P110"],

    "Age": [45, 62, 34, 55, 41, 68, 29, 52, 47, 73],

    "City": ["Hyderabad", "Warangal", "Hyderabad", "Vijayawada",
             "Hyderabad", "Nizamabad", "Hyderabad", "Warangal",
             "Hyderabad", "Karimnagar"],

    "Surgery_Type": [
        "Cholecystectomy", "CABG", "Appendectomy",
        "Hernia Repair", "Knee Replacement", "PTCA",
        "Appendectomy", "Cholecystectomy",
        "Hernia Repair", "Knee Replacement"
    ],

    "Operation_Duration_Min": [90, 240, 75, 110, 180,
                               150, 70, 95, 120, 200],

    "Length_of_Stay_Days": [2, 7, 1, 3, 5,
                            4, 1, 2, 3, 6],

    "Admission_Type": [
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Planned"
    ]
}

df = pd.DataFrame(data)

df
```

```python
# Unique() : means which is not similar to the other data, which tells the different values in the column . 
#  it returns the unique values in the specified column of the dataframe. it can be used to find the unique values in a column of a dataframe. 
# vlues deta hai
df["City"].unique()
```

```python
# nunique(): it returns the number of unique values in the specified column of dataframe. 
# it can be used to find the number of unique values in a column of a dataframe.

df["City"].nunique()
```

```python
df['City'].value_counts() #how many times the values is repeated in the column. it returns the count of unique values in the specified column of the dataframe. 
```

```python
df['Surgery_Type'].value_counts()
```

```python
# kaun kaun sa admission type hai aur kitne patients hai har admission type me.
df['Admission_Type'].unique()
```

```python
df['Admission_Type'].nunique() 
```

```python
df['Admission_Type'].value_counts()
```

```python
# standard function in pandas to get the unique values in a column of a dataframe is unique() and to get the number of unique values in a column of a dataframe is nunique().

# std() = data kitna faila hua hai, kitna variation hai data me.
#  it returns the standard deviation of the specified column of the dataframe.
#  it can be used to find the standard deviation of a column of a dataframe.

df["Operation_Duration_Min"].std()
```
