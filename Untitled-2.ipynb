```python
import pandas as pd

data = {
    "Student": ["Amit", "Neha", "Rahul", "Pooja"],
    "Marks": [85, 92, 76, 88],
    "City": ["Nagpur", "Pune", "Mumbai", "Nashik"]
}

df = pd.DataFrame(data, index=[101, 105, 110, 115])

df
```

```python
df.loc[110] 
```

```python
df.iloc[0 : 2] 
```

```python
# position 1 se 3 
df.iloc[1 : 3] 
```

```python
# iloc[start:stop] - This is used to select rows based on their integer position in the DataFrame. The start index is inclusive, and the stop index is exclusive. For example, df.iloc[1:3] will return rows at positions 1 and 2.(start included, stop excluded)
```

```python
# postion 1 to 3 se rows select karne ke liye hum iloc[1:3] ka use karte hai. Yaha 1 included hai aur 3 excluded hai. Iska matlab hai ki ye rows at positions 1 and 2 ko return karega.
df.iloc[1:3]
```
